Assignment:- Miniproject on rainfall data of the week

```python
# WAP to calculate Total, avg rain fall and rainy days for the week.
# rainfall = [0, 5, 12, 0, 0, 18, 25] in mm
# Days     = ["Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"]

import numpy as np
rainfall= np.array([0, 5, 12, 0, 0, 18, 25])
day= np.array(["Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"])
print("Rainfall:",rainfall)
print("Days:",day)

# Calculation of total, Avg, max, min rainfall of the week
total=np.sum(rainfall)
print("Total rainfall in this week:",total,"mm")
avg=np.mean(rainfall)
print("Avg rainfall in this week:",avg,"mm")
max=np.max(rainfall)
print("Maximum rainfall in this week:",max,"mm")
min=np.min(rainfall)
print("Minimum rainfall in this week:",min,"mm")

# Finding days having No rain, rainy & heavy rainy days
# More then 15mm rain per day is heavy rain
norain_days= day[rainfall==0]
print("Day's having no rain fall in this week:",norain_days)
rainy_days= day[rainfall>0]
print("Day's having rainfall in this week:",rainy_days)
heavyrainy_days= day[rainfall>15]
print("Day's having heavy rain fall in this week:",heavyrainy_days)
```
