(Problem Statement):-
A growing educational institution maintains thousands of books across multiple departments. 
Currently, the librarian records book information manually, making it difficult to search for books, track availability, 
update records, and maintain an organized catalog. 
As the number of books and students increases, the library requires a computerized solution to efficiently manage its collection.
You have been hired as a Python Developer to build a Library Management System using Python. 
The system should allow librarians to store, retrieve, update, and manage book information using appropriate Python programming concepts.
Your application should provide a menu-driven interface where users can perform various library operations such as adding new books, 
searching for books, issuing books, returning books, and displaying the complete catalog.

The application should be designed using functions, lists, dictionaries, loops, conditional statements, file handling (optional), and object-oriented programming (optional for advanced students) to demonstrate the practical application of Python programming concepts.

The Library Management System should allow the librarian to:
1.	Add a new book to the library. 
2.	Display all available books. 
3.	Search for a book using: 
o	Book ID 
o	Book Title 
o	Author Name 
4.	Update book details. 
5.	Delete a book from the library. 
6.	Issue a book to a student. 
7.	Return an issued book. 
8.	Display only available books. 
9.	Display issued books. 
10.	Exit the application safely.

```python
## Library Management System

library = []

# 1. Add a new book to the library:-
def add_book():
    book_id = input("Enter Book ID: ")

    for book in library:
        if book["id"] == book_id:
            print("Book ID already exists!")
            return

    title = input("Enter Book Title: ")
    author = input("Enter Author Name: ")

    book = {
        "id": book_id,
        "title": title,
        "author": author,
        "status": "Available",
        "student": ""
    }

    library.append(book)
    print("Book added successfully!")


# 2. Display all available books:-
def display_books():
    if not library:
        print("Library is empty!")
        return

    for book in library:
        print("----------------------")
        print("Book ID:", book["id"])
        print("Title:", book["title"])
        print("Author:", book["author"])
        print("Status:", book["status"])


# 3. Search for a book using(Book ID/Book Title/Author Name):-
def search_book():
    print("1. Search by Book ID")
    print("2. Search by Title")
    print("3. Search by Author")

    choice = input("Enter choice: ")
    keyword = input("Enter search value: ").lower()

    found = False

    for book in library:
        if choice == "1":
            match = book["id"].lower() == keyword
        elif choice == "2":
            match = keyword in book["title"].lower()
        elif choice == "3":
            match = keyword in book["author"].lower()
        else:
            print("Invalid choice!")
            return

        if match:
            print(book)
            found = True

    if not found:
        print("Book not found!")


# 4. Update book details:-
def update_book():
    book_id = input("Enter Book ID to update: ")

    for book in library:
        if book["id"] == book_id:
            book["title"] = input("Enter New Title: ")
            book["author"] = input("Enter New Author: ")

            print("Book updated successfully!")
            return

    print("Book not found!")


# 5. Delete a book from the library:-
def delete_book():
    book_id = input("Enter Book ID to delete: ")

    for book in library:
        if book["id"] == book_id:
            if book["status"] == "Issued":
                print("Cannot delete an issued book!")
                return

            library.remove(book)
            print("Book deleted successfully!")
            return

    print("Book not found!")


# 6. Issue a book to a student:-
def issue_book():
    book_id = input("Enter Book ID to issue: ")

    for book in library:
        if book["id"] == book_id:
            if book["status"] == "Available":
                student = input("Enter Student Name: ")

                book["status"] = "Issued"
                book["student"] = student

                print("Book issued successfully!")
            else:
                print("Book is already issued!")

            return

    print("Book not found!")


# 7. Return an issued book:-
def return_book():
    book_id = input("Enter Book ID to return: ")

    for book in library:
        if book["id"] == book_id:
            if book["status"] == "Issued":
                book["status"] = "Available"
                book["student"] = ""

                print("Book returned successfully!")
            else:
                print("Book is not issued!")

            return

    print("Book not found!")


# 8. Display only available books:-
def available_books():
    found = False

    for book in library:
        if book["status"] == "Available":
            print(book)
            found = True

    if not found:
        print("No available books!")


# 9. Display issued books:-
def issued_books():
    found = False

    for book in library:
        if book["status"] == "Issued":
            print(book)
            found = True

    if not found:
        print("No issued books!")


# 10. Exit the application safely:-
def main():
    while True:
        print("\n<<<<<<<< LIBRARY MANAGEMENT SYSTEM >>>>>>>>")
        print("1. Add New Book")
        print("2. Display All Books")
        print("3. Search Book")
        print("4. Update Book")
        print("5. Delete Book")
        print("6. Issue Book")
        print("7. Return Book")
        print("8. Display Available Books")
        print("9. Display Issued Books")
        print("10. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            add_book()

        elif choice == "2":
            display_books()

        elif choice == "3":
            search_book()

        elif choice == "4":
            update_book()

        elif choice == "5":
            delete_book()

        elif choice == "6":
            issue_book()

        elif choice == "7":
            return_book()

        elif choice == "8":
            available_books()

        elif choice == "9":
            issued_books()

        elif choice == "10":
            print("Thank you for using Library Management System!")
            break

        else:
            print("Invalid choice! Please try again.")
```

```python
# Start the program:-
main()
```
